```python
city_revenue = df.groupby(by="City")["Revenue"].sum()
city_revenue.plot(kind="bar")
plt.title("City-wise Total Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
cuisine_count =df["Cuisine"].value_counts()
cuisine_count.plot(kind="bar")
plt.title("Cuisine-wise Restaurant Count")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurant")
plt.xticks(rotation=45)
plt.show()
```

```python
#Rating Distribution

plt.hist(df["Rating"], bins=10)
plt.title("Restaurant Rating Destribution 1")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
online_order_count = df["Online_Order"].value_counts()
online_order_count.plot(kind="bar")
plt.title("Online order Avaliability")
plt.xlabel("Online Order")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
#Average Delivery Time by city
city_delivery = df.groupby(by="City")["Delivery_Time"].mean()
city_delivery.plot(kind="bar")
plt.title("Average Delivery Time by City")
plt.xlabel("City")
plt.ylabel("Average Delivery Time(Minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
#Orders vs Revenue
plt.scatter(df["Total_Orders"],
df["Revenue"])
plt.title("total orders vs Revenue")
plt.xlabel("Total Orders")
plt.ylabel("Revenue")
plt.show()
```

```python
#Rating vs Delivert time
plt.scatter(df["Delivery_Time"],
df["Rating"])
plt.title("Rating vs Delivery Time")
plt.xlabel("Delivery Time")
("Minutes)")
plt.ylabel("Rating")
plt.show()
```

```python
city_revenue = df.groupby(by="City")["Revenue"].sum()
city_revenue = city_revenue.sort_values(ascending=False)
print(city_revenue)
```

```python
#Top Cuisine by Orders
cuisine_orders = df.groupby(by="Cuisine")["Total_Orders"].sum()

cuisine_orders = cuisine_orders.sort_values(ascending=False)

print(cuisine_orders)
```
